```python
import math
import time
```

```python
def funcion(x):
    return math.sqrt(x) + x**2 + math.sin(x) + math.cos(x) + math.log(x)
```

```python
N = 2_000_000

datos = [i / 100 + 1 for i in range(N)]

print(f"Cantidad de datos: {len(datos)}")
print(f"Primer dato: {datos[0]}")
print(f"Último dato: {datos[-1]}")
```

```python
tiempos_secuenciales = []

for i in range(3):
    inicio = time.perf_counter()

    resultados = [funcion(x) for x in datos]

    fin = time.perf_counter()

    tiempo = fin - inicio
    tiempos_secuenciales.append(tiempo)

    print(f"Ejecución {i + 1}: {tiempo:.6f} segundos")

promedio_secuencial = sum(tiempos_secuenciales) / len(tiempos_secuenciales)

print(f"\nPromedio secuencial: {promedio_secuencial:.6f} segundos")
```
