## ¿Qué es HPC?

**HPC (High Performance Computing / Computación de Alto Rendimiento)** es el área que estudia cómo resolver problemas de cómputo que son demasiado grandes o demasiado lentos para una sola computadora personal.

Piensa en la diferencia entre:
- Correr un script de Python en tu laptop que tarda 2 minutos.
- Simular el clima de todo el planeta para mañana, entrenar un modelo con terabytes de datos, o procesar las imágenes de un telescopio — cosas que en una laptop tardarían años.


## Clústeres y vocabulario de HPC

Para el segundo tipo de problema se usan **clústeres**: cientos o miles de computadoras conectadas por una red muy rápida. Este es el vocabulario clave que vas a usar durante todo el curso:

| Término | Significado |
|---|---|
| Nodo | Una "computadora" dentro del clúster (tiene su propia memoria RAM y CPU) |
| Core / núcleo | Una unidad de procesamiento dentro de un CPU; hoy un CPU típico tiene entre 8 y 128 cores |
| Cluster / clúster | Conjunto de nodos conectados en red, administrados como un solo sistema |


## GPU, paralelismo y MPI

| Término | Significado |
|---|---|
| GPU | Procesador diseñado para hacer miles de operaciones simples en paralelo (útil en IA y simulación) |
| Paralelismo | Dividir un problema en partes que se resuelven al mismo tiempo, en distintos cores o nodos |
| MPI | Estándar para que procesos en distintos nodos se comuniquen entre sí (lo verás en el notebook 13) |

## Python como capa de control

Es cierto que un `for` puro en Python es lento comparado con C. Pero en la práctica de HPC casi nunca se escribe cómputo pesado en Python puro: se usa Python como capa de control sobre librerías escritas en C/C++/Fortran (NumPy, SciPy), o se acelera el propio código Python con compiladores JIT (Numba, JAX) o ejecutándolo en GPU. Ese es el hilo conductor de todo el curso: escribir Python de la manera correcta para HPC.


## Ejemplos de códigos de simulación HPC completos hechos en Python

Estrategia común en software científico serio: el núcleo pesado del cálculo está escrito en C/C++, y Python se usa como interfaz para controlarlo, combinarlo y analizarlo. Dos ejemplos reales:


### [GPAW](https://wiki.fysik.dtu.dk/gpaw/)

GPAW simula cómo se comportan los electrones dentro de un material o una molécula, usando un método llamado *Density Functional Theory* (DFT, "teoría del funcional de la densidad"). En términos simples: en vez de hacer un experimento físico en un laboratorio para saber si un material nuevo conduce electricidad, es estable, o reacciona con otra sustancia, se simula a nivel cuántico en una computadora. Se usa en ciencia de materiales, diseño de baterías, catalizadores, semiconductores, etc.


## GPAW: cálculo y paralelismo

* El cálculo (resolver ecuaciones cuánticas para millones de puntos) está escrito en C, porque necesita ser muy rápido.
* Tú, como usuario, escribes un script corto en Python describiendo qué material quieres simular.
* Se paraleliza con MPI + OpenMP y puede escalar a decenas de miles de cores en un supercómputo.


### [ESPResSo++](http://www.espresso-pp.de)

ESPResSo++ simula dinámica molecular: cómo se mueven en el tiempo miles o millones de átomos o moléculas bajo fuerzas físicas (atracción, repulsión, colisiones). Se especializa en "materia blanda" — polímeros, plásticos, cadenas largas de moléculas parecidas al ADN. Es la herramienta que usarías si quisieras simular, por ejemplo, cómo se enredan las cadenas de un plástico al enfriarse.


## ESPResSo++: cálculo y paralelismo

* El motor de simulación (calcular fuerzas entre millones de partículas, paso a paso en el tiempo) está en C++.
* Tú controlas la simulación (qué partículas, qué fuerzas, cuántos pasos) desde un script de Python.
* Se paraleliza dividiendo el espacio de simulación en regiones ("descomposición de dominio") y usando MPI para que cada región se calcule en un proceso distinto.


## El patrón común en HPC

**El patrón que debes quedarte de estos dos ejemplos:** en HPC casi nunca vas a escribir *todo* en Python. Vas a usar Python para dirigir cálculos que corren en C/C++/Fortran, en GPU, o en muchos procesos a la vez. Ese es exactamente el tipo de habilidad que vas a construir en este curso: no solo "saber Python", sino saber cuándo y cómo conectarlo con herramientas más rápidas.

## Por qué usar Miniforge

Son bastantes paquetes, y varios de ellos (mpi4py, h5py con soporte MPI, los compiladores) necesitan versiones compiladas de forma muy específica para funcionar bien juntos. Instalarlos uno por uno con `pip` suele terminar en errores difíciles de depurar. Por eso vamos a usar **Miniforge**.


## Tarea 1: configura tu entorno

### ¿Qué es un entorno virtual y por qué lo necesitas?

Si instalas paquetes de Python directamente en tu sistema, tarde o temprano dos proyectos van a necesitar versiones distintas (o incompatibles) de la misma librería, y uno de los dos deja de funcionar. Un **entorno virtual** es una instalación de Python aislada, propia de un proyecto: instalas ahí las versiones exactas que ese proyecto necesita, sin afectar nada más en tu computadora ni depender de lo que ya tengas instalado.


### ¿Qué es Miniforge?

**Conda** es un gestor de entornos y paquetes. A diferencia de `pip`, conda no solo maneja paquetes de Python: también maneja binarios compilados (compiladores, librerías de MPI, BLAS, HDF5, etc.), que es justo lo que necesita este curso. Eso importa mucho en HPC: paquetes como `mpi4py` o `h5py` con soporte MPI necesitan compilarse contra versiones específicas de librerías de sistema, y conda garantiza que todas las piezas encajen entre sí.


## Miniforge y conda-forge

**Miniforge** es la forma recomendada hoy en día de instalar conda:
* Es un instalador mínimo (no viene con cientos de paquetes preinstalados como Anaconda clásico).
* Usa por defecto el canal **conda-forge**: un repositorio de paquetes mantenido por la comunidad, gratuito y sin restricciones de licencia para uso académico/comercial (a diferencia del canal por defecto de Anaconda Inc.).
* Incluye `mamba`, una versión más rápida de conda para resolver e instalar dependencias.

### Aviso para usuarios de macOS y Windows

El archivo `environment.yml` incluye los paquetes `gcc_linux-64`, `gxx_linux-64` y `gfortran_linux-64`. Estos son compiladores empaquetados **específicamente para Linux**; en macOS o Windows nativo el paso `conda env create` va a fallar en esa parte.


## Crear el entorno en macOS o Windows

Si usas **macOS o Windows**, usa en su lugar el archivo `environment2.yml` (incluido en este repositorio), que reemplaza esos tres paquetes por `compilers`, el metapaquete de conda-forge que elige automáticamente el compilador correcto según tu sistema operativo (clang en macOS, gcc en Linux):

```bash
conda env create --file environment2.yml
conda activate pyhpc
```

Si usas **Linux**, usa el `environment.yml` original tal como está.
